## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code

import ast
import matplotlib.pyplot as plt

# Filter the dataframe for type == "SHOW"
shows_df = df[df["type"] == "SHOW"]

# Extract and count genres
genre_counts = {}
for genres in shows_df["genres"].dropna():
    genre_list = ast.literal_eval(genres)  # Convert string to list
    for genre in genre_list:
        genre_counts[genre] = genre_counts.get(genre, 0) + 1

# Sort and plot the genre counts
plt.figure(figsize=(12, 6))
plt.bar(genre_counts.keys(), genre_counts.values(), color='skyblue')
plt.xlabel("Genres")
plt.ylabel("Count of Shows")
plt.title("Number of Shows by Genre")
plt.xticks(rotation=45, ha="right")
plt.show()



````

##### 1. Why did you pick the specific chart?

To have an idea about the distribution of genres based on shows present.

##### 2. What is/are the insight(s) found from the chart?

The majority of shows fall under the drama genre, followed by comedy. There are over 300 animated shows. Additionally, more than 200 shows belong to the action, documentary, family, and sci-fi genres. Crime, European, fantasy, reality, and romance genres each have over 100 shows. The remaining genres have significantly lower counts, with western being the least represented.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, drama and comedy are generally the most preferred genres, leading to higher subscription counts and driving positive growth.

#### Chart - 2

````python
# Chart - 2 visualization code
import ast
import matplotlib.pyplot as plt

# Filter movies only
movies_df = df[df["type"] == "MOVIE"]

# Extract and count genres
genre_counts = {}
for genres in movies_df["genres"]:
    genre_list = ast.literal_eval(genres)  # Convert string representation of list to actual list
    for genre in genre_list:
        genre_counts[genre] = genre_counts.get(genre, 0) + 1

# Sort genres by count
sorted_genres = dict(sorted(genre_counts.items(), key=lambda x: x[1], reverse=True))

# Plot bar chart
plt.figure(figsize=(12, 6))
plt.bar(sorted_genres.keys(), sorted_genres.values(), color='skyblue')
plt.xticks(rotation=45, ha='right')
plt.xlabel("Genres")
plt.ylabel("Number of Movies")
plt.title("Number of Movies by Genre")
plt.show()

````

#### Chart - 5

````python
# Chart - 5 visualization code
import ast
import matplotlib.pyplot as plt

# Filter dataset to include only shows
shows_df = df[df["type"] == "SHOW"]

# Function to clean and extract country names from string representation of lists
def extract_countries(country_str):
    try:
        return ast.literal_eval(country_str) if isinstance(country_str, str) else []
    except:
        return []

# Apply extraction function
shows_df["production_countries"] = shows_df["production_countries"].apply(extract_countries)

# Flatten the list and count occurrences
country_counts = pd.Series([country for sublist in shows_df["production_countries"] for country in sublist]).value_counts()

# Plot the bar chart
plt.figure(figsize=(12, 6))
country_counts.head(10).plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Top 10 Production Countries for Shows")
plt.xlabel("Country")
plt.ylabel("Number of Shows")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
````

##### 1. Why did you pick the specific chart?

 It allows direct comparison of the top 10 countries, highlighting disparities in production volume.

##### 2. What is/are the insight(s) found from the chart?

Countries with high production numbers have strong entertainment industries, while those with fewer indicate emerging markets. Streaming platforms can use this data for content licensing and production, shaping international acquisition strategies based on dominant countries.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Businesses can prioritize content deals with high-production countries, optimize localization efforts, and identify key markets for expansion. Focusing on underserved markets can provide a competitive advantage with unique content offerings.

#### Chart - 6

````python
# Chart - 6 visualization code
import ast
import matplotlib.pyplot as plt

# Filter only movies
movies_df = df[df['type'] == 'MOVIE'].copy()

# Function to convert string representation of list to actual list
def parse_countries(country_str):
    try:
        return ast.literal_eval(country_str) if isinstance(country_str, str) else []
    except (ValueError, SyntaxError):
        return []

# Extract and count production countries
movies_df['production_countries'] = movies_df['production_countries'].apply(parse_countries)
all_countries = [country for sublist in movies_df['production_countries'] for country in sublist]
country_counts = pd.Series(all_countries).value_counts()

# Plot the distribution
plt.figure(figsize=(12, 6))
country_counts[:10].plot(kind='bar', color='skyblue')  # Top 10 countries
plt.title("Top 10 Production Countries for Movies")
plt.xlabel("Country")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)
plt.show()
````

##### 1. Why did you pick the specific chart?

Focus on Top 10: Highlighting only the top 10 avoids clutter and keeps the analysis relevant.

##### 2. What is/are the insight(s) found from the chart?
